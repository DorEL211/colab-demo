# Creative Ideas: Three Uses for an Everyday Object

Enter an everyday object and get three unusual, practical ideas. Keep entering objects to explore more; type 'exit' or 'quit' when you are done. Run the cells from top to bottom.

```python
# Install the OpenAI Python SDK in this Colab session.
%pip install -q openai
```

```python
from getpass import getpass
from openai import OpenAI

# Ask for your key once, without showing it as you type.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

# Keep asking for objects until the user chooses to stop.
while True:
    object_name = input("Name an everyday object (or type 'exit' to stop): ").strip()
    if object_name.lower() in ("exit", "quit"):
        print("Goodbye!")
        break
    if not object_name:
        print("Please enter an object, or type 'exit' to stop.")
        continue

    # Ask for exactly three numbered, practical ideas.
    response = client.responses.create(
        model="gpt-6-luna",
        input=(
            f"For the everyday object {object_name!r}, suggest exactly THREE unusual, practical uses. "
            "Make them safe, realistic, beginner-friendly, and distinct. "
            "Format the answer as a numbered list from 1 to 3, with a brief explanation for each idea."
        ),
    )
    print(response.output_text)
    print()
```
