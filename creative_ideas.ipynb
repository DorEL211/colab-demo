# Creative Ideas: One New Use for an Everyday Object

Enter an everyday object and let the OpenAI Responses API suggest one unusual, practical way to use it. Run the cells from top to bottom.

```python
# Install the OpenAI Python SDK in this Colab session.
%pip install -q openai
```

```python
from getpass import getpass
from openai import OpenAI

# Ask for your key without showing it as you type.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

# Ask for an everyday object.
object_name = input("Name an everyday object: ").strip()
if not object_name:
    raise ValueError("Please enter an object and run this cell again.")
```

```python
# Request exactly one unusual but practical idea.
response = client.responses.create(
    model="gpt-6-luna",
    input=(
        f"Suggest exactly ONE unusual, practical use for this everyday object: {object_name}. "
        "Keep it safe, realistic, and beginner-friendly. Explain the idea in 2-3 sentences."
    ),
)

print(response.output_text)
```
